```python
c = CurrencyRates()

@lru_cache(maxsize=1000)
def _get_usd_to_inr_rate_for_date(date: datetime.date):
    """
    Fetches the USD to INR exchange rate for a specific date using forex-python.
    Caches the result to avoid repeated API calls for the same date.
    
    The date parameter should be a datetime.date object.
    For the latest rate, pass the current date.
    """
    try:
        # get_rate expects a datetime object for historical rates
        datetime_obj = datetime.combine(date, datetime.min.time())
        return c.get_rate('USD', 'INR', datetime_obj)
    except Exception as e:
        print(f"Error fetching rate for {date}: {e}")
        # Fallback to a fixed rate or raise an error as appropriate
        # For demonstration, we'll return a placeholder.
        return 83.0 # Fallback rate if API fails

def get_latest_usd_to_inr_rate():
    """Fetches the latest USD to INR exchange rate."""
    return c.get_rate('USD', 'INR')

# The original lists remain the same
_MONETARY_INFO_KEYS = [
    'MarketCap', 'TotalAssets', 'TotalRevenue', 'EBITDA', 'RegularMarketOpen',
    'RegularMarketDayHigh', 'RegularMarketDayLow', 'RegularMarketPreviousClose',
    'RegularMarketPrice', 'FiftyTwoWeekHigh', 'FiftyTwoWeekLow', 'BookValue'
]
_MONETARY_DF_COLS = ['Open', 'High', 'Low', 'Close', 'Adj Close', 'Dividends']


class InrTicker:
    """A wrapper for yf.Ticker that safely converts USD data to INR."""

    def __init__(self, symbol):
        self.symbol = symbol
        self._ticker = yf.Ticker(self.symbol)
        self.original_info = self._ticker.info
        self.info = self._convert_info()

    def _convert_info(self):
        info_copy = self.original_info.copy()
        
        # Get the latest exchange rate for current fundamental data
        usd_to_inr_rate = get_latest_usd_to_inr_rate()
        
        if info_copy.get('currency', '').upper() == 'USD':
            for key in _MONETARY_INFO_KEYS:
                # Use float/int check for safety, and check if the key exists
                if key in info_copy and isinstance(info_copy.get(key), (int, float)):
                    info_copy[key] *= usd_to_inr_rate
                    info_copy[f'{key}_inr'] = info_copy[key] # New INR field
            
            info_copy['currency'] = 'INR'
            info_copy['original_currency'] = 'USD' # Keep track of original currency
            
        return info_copy

    def history(self, *args, **kwargs):
        df = self._ticker.history(*args, **kwargs)
        
        # Check if the data is already converted or if conversion is needed
        # We assume the original data is in USD if the history call is made
        # and the ticker is a USD ticker (though the currency check on self.info is better)
        if self.original_info.get('currency', '').upper() == 'USD' and not df.empty:
            
            # 1. Prepare an array of daily conversion rates
            # df.index is a DatetimeIndex, convert to date objects for the function
            dates_to_fetch = df.index.normalize().map(lambda x: x.date())
            daily_rates = [
                _get_usd_to_inr_rate_for_date(date) for date in dates_to_fetch
            ]
            rate_series = pd.Series(daily_rates, index=df.index)

            # 2. Apply the daily rate to the monetary columns
            for col in _MONETARY_DF_COLS:
                if col in df.columns:
                    # Multiply the column data by the corresponding daily rate
                    df[col] *= rate_series
                    
        return df

def yf_inr_download(tickers, *args, **kwargs):
    """
    A wrapper for yf.download that safely converts USD data to INR using daily rates.
    """
    
    # # 1. Download original data first
    data = yf.download(tickers, *args, **kwargs)
    if data.empty:
        return data

    # # 2. Find which tickers are in USD (simplified for this example, assumes single ticker if not multi-index)
    # The original implementation had a loop to check currency, we will use a simpler check 
    # and assume that if the function is used, conversion is desired for known USD tickers.
    
    # In a real-world scenario, you would need to fetch info for each ticker as in the original code's step 2
    # For a direct modification of the provided code, we'll keep the loop structure (with a placeholder for simplicity)
    
    tickers_list = tickers.replace(',', ' ').split()
    usd_tickers = []
    for ticker_symbol in tickers_list:
        try:
            # We must use yf.Ticker to get the currency
            info = yf.Ticker(ticker_symbol).info
            if info.get('currency', '').upper() == 'USD':
                usd_tickers.append(ticker_symbol)
        except Exception:
            pass # Ignore tickers that can't be found

    if not usd_tickers:
        return data # Return original data if no USD tickers to convert

    # # 3. Convert only the USD tickers
    print(f"Converting data to INR for: {', '.join(usd_tickers)}")

    # Prepare an array of daily conversion rates based on the DataFrame's index (dates)
    dates_to_fetch = data.index.normalize().map(lambda x: x.date())
    daily_rates = [
        _get_usd_to_inr_rate_for_date(date) for date in dates_to_fetch
    ]
    rate_series = pd.Series(daily_rates, index=data.index)
    
    is_multi_index = isinstance(data.columns, pd.MultiIndex)

    for ticker_symbol in usd_tickers:
        for col in _MONETARY_DF_COLS:
            if is_multi_index:
                # Handle multi-ticker dataframes (e.g., data['Open', 'TSLA'])
                if (col, ticker_symbol) in data.columns:
                    # Multiply the column by the daily rate series
                    data[col, ticker_symbol] *= rate_series
            else:
                # Handle single-ticker dataframes (col is just 'Open')
                if len(tickers_list) == 1 and col in data.columns:
                    # Multiply the column by the daily rate series
                    data[col] *= rate_series

    return data
```

```python
from sklearn.covariance import LedoitWolf

def get_shrunk_covariance(returns):
  lw_estimator = LedoitWolf()
  lw_estimator.fit(returns)
  shrunk_covariance = lw_estimator.covariance_
  return shrunk_covariance
column_headers = returns.columns
shrunk_returns = get_shrunk_covariance(returns)
shrunk_df = pd.DataFrame(shrunk_returns, 
                          index=column_headers, 
                          columns=column_headers)
shrunk_df
```

```python
manual_aum = {
    'ITBEES.NS': 2.438e+10,
    'GOLDBEES.NS': 2.2355e+11,  
    'SILVERBEES.NS': 8.078e+10,
    'BBETF0432.NS' : 1.073e+11,
    'BBETF0432.NS' : 1.073e+11,
    'NETFDIVOPP.NS' : 7.547e+8,
    'HEALTHY.NS' : 3.657e+8,
    'NIFTYBEES.NS' : 4.892e+11
}

num = np.arange(len(asset_symbols))
market_cap = np.zeros(len(asset_symbols))

for i in num:
    symbol = asset_symbols[i]
    if symbol in manual_aum:
        market_cap[i] = manual_aum[symbol]
        print(f"Using manual AUM for {symbol}")
        continue
    ticker = InrTicker(symbol)
    
    assets = ticker.info.get('totalAssets', None)
    m_cap = ticker.info.get('marketCap', None)
    
    if assets is not None:
        market_cap[i] = assets
    elif m_cap is not None:
        market_cap[i] = m_cap
    else:
        market_cap[i] = 0
        print(f"Warning: Could not find data for {symbol}")

print("\nFinal Data Array:")
print(market_cap)
```

```python
# 1. Calculate market-cap weights
market_weights = market_cap / market_cap.sum()

# 2. Calculate the market's risk aversion (delta)
# A common method is to use the market's Sharpe Ratio.
market_returns_annual = (1 + returns.mean()).prod()**252 - 1
market_vol_annual = returns.std() * np.sqrt(252)
# Assuming a simple equal-weighted market portfolio for Sharpe calculation
sharpe_ratio_market = np.mean(market_returns_annual) / np.mean(market_vol_annual)
risk_aversion = sharpe_ratio_market / np.mean(market_vol_annual) # delta

# 3. Calculate the final implied returns vector (Pi)
implied_returns = risk_aversion * shrunk_df.dot(market_weights)

print("--- Annual Market-Implied Returns (Π) ---")
print((implied_returns * 100).round(2).to_string())
```

```python
q = np.array([
    0.076,  
    0.07   
])[:, np.newaxis]

p = np.zeros((len(q), len(returns.columns)))
p[0, 5] = 1.0  
p[0, 14] = 1.0  
p[1, 24] = -1.0 
tau = 0.02
omega = np.diag(np.diag(tau * p @ shrunk_df @ p.T))
p
```

```python
# The Black-Litterman formula for the posterior expected returns
bl_part_1 = np.linalg.inv(np.linalg.inv(tau * shrunk_df) + p.T @ np.linalg.inv(omega) @ p)
bl_part_2 = np.linalg.inv(tau * shrunk_df) @ implied_returns + p.T @ np.linalg.inv(omega) @ q.flatten()
black_litterman_returns = bl_part_1 @ bl_part_2

# Present the final returns nicely
bl_returns_df = pd.DataFrame({'Implied': implied_returns, 'Black-Litterman': black_litterman_returns},
                             returns.columns)

print("\n--- Final Blended Expected Returns (in %) ---")
print(bl_returns_df.mul(100).round(2))
```

```python
import pandas as pd
import cvxpy as cp

# Add 'asset_names' as a new argument
def optimize_portfolio(returns_np, shrunk_df, risk_aversion, asset_names):
    
    num_assets = len(returns_np)
    weights = cp.Variable(num_assets)
    
    expected_return = weights @ returns_np
    risk = cp.quad_form(weights, shrunk_df)
    
    objective = cp.Maximize(expected_return - 0.5 * risk_aversion * risk)
    constraints = [cp.sum(weights) == 1]
    constraints.append(weights >= 0)
    
    problem = cp.Problem(objective, constraints)
    problem.solve()
    
    if problem.status == 'optimal':
        # Use the 'asset_names' list you passed in
        return pd.Series(weights.value, index=asset_names)
    else:
        return None

# Get the asset names from your original returns DataFrame/Series
asset_names = returns.columns

# Call the updated function with the asset names as the last argument
optimal_weights = optimize_portfolio(
    black_litterman_returns,
    shrunk_df,
    risk_aversion,
    asset_names  # <-- Pass the labels here
)

print(optimal_weights)
```
